```python
import xtrack as xt

base = "https://acc-models.web.cern.ch/acc-models/lhc/hl19"
lhc = xt.load(f"{base}/xsuite/lhc.json", timeout=60)
lhc.vars.load( f"{base}/strengths/cycle_round_v0/ramp/ramp_A600cmC600cmA10mL10m_0.madx")
lhc.vars.load(f"{base}/scenarios/cycle/pp/ramp/0/settings.madx")

lhc.aperture.from_json(f"{base}/aperture/aperture.json")
```

```python
aperture_b1 = lhc.b1.aperture
aperture_b1.build_model()
s_pos_b1 = aperture_b1.get_s_around_transitions()
# Widen the two-panel extent figure to use the notebook output area.
fig_b1, axes_b1 = aperture_b1.plot_extents(s_pos_b1)
fig_b1.set_size_inches(16, 8)
```

```python
aperture_b2 = lhc.b2.aperture
aperture_b2.build_model()
s_pos_b2 = aperture_b1.get_s_around_transitions()
# Match the width used for Beam 1 so the two plots are easy to compare.
fig_b2, axes_b2 = aperture_b2.plot_extents(s_pos_b2)
fig_b2.set_size_inches(16, 8)
```

```python
# Zoom in around CMS
s_pos_cms = s_pos_b1[(13_000 < s_pos_b1) & (s_pos_b1 < 13_650)]
fig_b1, axes_b1 = aperture_b1.plot_extents(s_pos_cms)
fig_b1.set_size_inches(16, 8)
```
